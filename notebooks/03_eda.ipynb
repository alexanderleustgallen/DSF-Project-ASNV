## 1. Tm distribution, overall and per organism

```python
order = df.groupby("organism")["tm"].median().sort_values().index
fig, ax = plt.subplots(figsize=(10, 5))
df.boxplot(column="tm", by="organism", positions=range(len(order)), ax=ax, rot=90)
ax.set_xticks(range(len(order)), order, rotation=90)
ax.set_ylabel("Tm (°C)")
plt.suptitle("")
plt.tight_layout()
```
