## 1. Baselines

Every other model has to beat these. `baseline_ogt` uses only the growth temperature, no sequence information.

```python
mean_model = DummyRegressor(strategy="mean").fit(X_train, y_train)
print(save_results("baseline_mean", y_test, mean_model.predict(X_test)))

ogt_model = LinearRegression().fit(X_train[["ogt"]], y_train)
print(save_results("baseline_ogt", y_test, ogt_model.predict(X_test[["ogt"]])))
```
